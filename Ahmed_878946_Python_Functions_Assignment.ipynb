## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
# Write your code below
def process_order(price, quantity):

    def calculate_subtotal():
        return price * quantity

    def calculate_discount():
        subtotal = calculate_subtotal()
        return subtotal * 0.10

    def calculate_tax():
        subtotal = calculate_subtotal()
        discount = calculate_discount()
        return (subtotal - discount) * 0.05

    def calculate_final_amount():
        subtotal = calculate_subtotal()
        discount = calculate_discount()
        tax = calculate_tax()

        return subtotal - discount + tax
        
    return calculate_final_amount()

price = 2000
quantity = 5

total = process_order(price, quantity)

print("Price per Item:", price)
print("Quantity:", quantity)
print("Final Order Amount:", total)

````

## Question 27 — Employee Payroll System

**Problem:**  
Build a small employee payroll system using multiple functions.

**Requirements:**
Your program should contain **separate functions** for:
- Employee information
- Basic salary
- Allowance
- Deduction
- Tax
- Final salary calculation

**Rules:**
- Use arguments and `return` properly.
- Use conditions where required (e.g., tax slabs).
- Do **not** write everything inside one function.
- Display a clean final salary report.

**Focus:** Function design + modular programming

````python
# Write your code below
def employee_info():
    name = input("Enter employee name: ")
    emp_id = input("Enter employee ID: ")
    return name, emp_id

def basic_salary():
    salary = float(input("Enter basic salary: "))
    return salary

def allowance(salary):
    house_allowance = salary * 0.20
    medical_allowance = salary * 0.10
    total_allowance = house_allowance + medical_allowance
    return total_allowance

def deduction(salary):
    provident_fund = salary * 0.05
    return provident_fund
    
def tax(salary):
    if salary <= 50000:
        tax_amount = 0
    elif salary <= 100000:
        tax_amount = salary * 0.05
    else:
        tax_amount = salary * 0.10

    return tax_amount

def final_salary(salary, allow, deduct, tax_amount):
    net_salary = salary + allow - deduct - tax_amount
    return net_salary
    
name, emp_id = employee_info()

salary = basic_salary()

allow = allowance(salary)

deduct = deduction(salary)

tax_amount = tax(salary)

net_salary = final_salary(salary, allow, deduct, tax_amount)

print("\n========== EMPLOYEE SALARY REPORT ==========")
print("Employee Name:", name)
print("Employee ID:", emp_id)
print("--------------------------------------------")
print("Basic Salary:       Rs.", salary)
print("Total Allowance:    Rs.", allow)
print("Total Deduction:    Rs.", deduct)
print("Income Tax:         Rs.", tax_amount)
print("--------------------------------------------")
print("Final Net Salary:   Rs.", net_salary)
print("============================================")

````

## Question 28 — Smart Shopping Billing System

**Problem:**  
Build a shopping billing system that can handle multiple products.

**Requirements:**
- Allow the user (or hard-code) multiple products.
- Calculate:
  - Individual product totals
  - Overall subtotal
  - Discount
  - Tax
  - Final bill
- Use **multiple functions** to separate responsibilities.
- At least one function must use `*args` or a flexible-argument design.
- Display a professional-looking bill.

**Focus:** Real-world application + multiple functions + logic building

````python
# Write your code below

def get_products():
    products = []
    n = int(input("How many products do you want to buy? "))

    for i in range(n):
        name = input("Enter product name: ")
        price = float(input("Enter product price: "))
        quantity = int(input("Enter quantity: "))
        total = product_total(price, quantity)
        products.append((name, price, quantity, total))

    return products


def product_total(price, quantity):
    return price * quantity


def subtotal(*prices):
    return sum(prices)


def discount(total):
    if total >= 5000:
        return total * 0.10
    else:
        return 0


def tax(amount):
    return amount * 0.05


def final_bill(total, dis, t):
    return total - dis + t


products = get_products()

prices = [p[3] for p in products]

total = subtotal(*prices)
dis = discount(total)
t = tax(total - dis)
bill = final_bill(total, dis, t)

print("\n===== SMART SHOPPING BILL =====")

for p in products:
    print(p[0], "Price:", p[1], "Quantity:", p[2],
          "Total:", p[3])

print("--------------------------------")
print("Subtotal:", total)
print("Discount:", dis)
print("Tax:", t)
print("Final Bill:", bill)
print("================================")
print("Thank you for shopping!")
````

````python
# Write your code below

def customer_info():
    name = input("Enter customer name: ")
    address = input("Enter address: ")
    return name, address


def product_selection():
    products = []
    n = int(input("How many products do you want to buy? "))

    for i in range(n):
        name = input("Enter product name: ")
        price = float(input("Enter product price: "))
        qty = int(input("Enter quantity: "))

        total = price * qty

        products.append({
            "name": name,
            "price": price,
            "qty": qty,
            "total": total
        })

    return products


def subtotal(*prices):
    return sum(prices)


def discount(total):
    if total >= 5000:
        return total * 0.10
    else:
        return 0


def tax(amount, rate=0.05):
    return amount * rate


def delivery(total):
    if total >= 3000:
        return 0
    else:
        return 200


def final_bill(total, dis, t, delivery_charge):
    return total - dis + t + delivery_charge


def order_summary(name, address, products, total, dis, t, charge, bill):

    print("\n====== E-COMMERCE ORDER SUMMARY ======")
    print("Customer Name:", name)
    print("Address:", address)
    print("--------------------------------------")

    for p in products:
        print(p["name"], "Price:", p["price"],
              "Qty:", p["qty"], "Total:", p["total"])

    print("--------------------------------------")
    print("Subtotal:", total)
    print("Discount:", dis)
    print("Tax:", t)
    print("Delivery Charges:", charge)
    print("--------------------------------------")
    print("Final Bill:", bill)
    print("Thank you for shopping!")
    print("======================================")


name, address = customer_info()

products = product_selection()

prices = [p["total"] for p in products]

total = subtotal(*prices)

dis = discount(total)

t = tax(total - dis)

charge = delivery(total)

bill = final_bill(total, dis, t, charge)

order_summary(name, address, products, total, dis, t, charge, bill)
````
